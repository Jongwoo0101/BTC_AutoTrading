```python
import pandas as pd
import pyupbit
from prophet import Prophet
```

```python
# 파라미터 조정 x
df = pyupbit.get_ohlcv("KRW-BTC", interval="minute60")
df = df.reset_index()
df['ds'] = df['index']
df['y'] = df['close']
data = df[['ds','y']]
m = Prophet()
m.fit(data)
```

```python
# 100일의 실적 예측
future = m.make_future_dataframe(periods=100)
forecast = m.predict(future)
```

```python
# 시각화
fig1 = m.plot(forecast)
```

```python
fig2 = m.plot_components(forecast)
```
